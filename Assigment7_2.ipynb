```python

# ============================================================
# Assignment 7.2: Clustering Sales Data Using K-Means
# AIT 620 | Activity 5.01 | Chapter 5
# ============================================================

from IPython.display import display, Markdown
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


def section(title, text):
    display(Markdown(f"## {title}\n\n{text}"))


pd.set_option("display.max_columns", 12)
pd.set_option("display.width", 120)

RANDOM_STATE = 8

# ============================================================
# ASSIGNMENT TITLE AND OBJECTIVE
# ============================================================

display(Markdown("""
# Assignment 7.2: Clustering Sales Data Using K-Means

**Course:** AIT 620, Methods of Artificial Intelligence and Machine Learning

**Activity:** 5.01

**Topic:** Chapter 5 — Clustering

**Dataset:** Sales Transactions Dataset Weekly

## Objective

Group products using 52 weeks of sales, reproduce the textbook's
eight-cluster analysis, and evaluate alternative numbers of clusters.

The dataset contains product identifiers, weekly sales figures,
and additional normalization-related fields.
"""))


# ============================================================
# 1. IMPORT THE LIBRARIES
# ============================================================

section(
    "1. Import the libraries",
    """
Pandas handles data loading and preprocessing. NumPy supports numerical
operations. Scikit-learn fits the K-Means clustering model and evaluates
cluster quality. Matplotlib produces charts to visualize the results.
"""
)

print("Libraries imported successfully.")
print("Random state:", RANDOM_STATE)


# ============================================================
# 2. LOAD AND INSPECT THE ORIGINAL DATASET
# ============================================================

section(
    "2. Load and inspect the original dataset",
    """
The dataset is downloaded directly from the provided GitHub repository.
The CSV file contains product identifiers, 52 weekly sales columns,
additional fields, and 52 normalized weekly features. We inspect the
dataset shape, column names, missing values, and first five records.
"""
)

file_path = (
    "https://raw.githubusercontent.com/"
    "PacktWorkshops/The-Applied-Artificial-Intelligence-Workshop/"
    "master/Datasets/Sales_Transactions_Dataset_Weekly.csv"
)

df = pd.read_csv(file_path)

print("Original dataset shape:", df.shape)
print("First eight columns:", df.columns[:8].tolist())
print("Last five columns:", df.columns[-5:].tolist())
print("Missing values:", int(df.isna().sum().sum()))

display(df.iloc[:5, :8])


# ============================================================
# 3. DROP THE 55 UNNECESSARY COLUMNS
# ============================================================

section(
    "3. Drop the 55 unnecessary columns",
    """
The original file has 107 columns: one product identifier, 52 weekly
sales columns, two MIN/MAX columns, and 52 normalized weekly features.

The activity instructs us to remove the first 55 columns, leaving the
52 normalized weekly features for clustering. We preserve the original
data in df so that the raw weekly sales can be used later to calculate
sales totals after assigning cluster labels.
"""
)

weekly_cols = [f"W{i}" for i in range(52)]

assert all(col in df.columns for col in weekly_cols), (
    "Some original weekly sales columns are missing."
)

columns_to_drop = df.columns[:55].tolist()

print("Columns to remove:", len(columns_to_drop))

sales_df = df.copy()
sales_df.drop(columns=columns_to_drop, inplace=True)

assert sales_df.shape == (811, 52), (
    f"Expected (811, 52), but received {sales_df.shape}"
)

assert sales_df.columns.tolist() == [
    f"Normalized {i}" for i in range(52)
], "The remaining columns do not match the expected normalized features."

print("Cleaned sales data shape:", sales_df.shape)
display(sales_df.head())


# ============================================================
# 4. FIT THE REQUIRED K-MEANS MODEL
# ============================================================

section(
    "4. Fit the required K-Means model",
    """
The assignment requires exactly eight clusters and random_state=8.
The model uses the 52 normalized weekly features. Setting n_init=10
runs ten centroid initializations to improve consistency across
scikit-learn versions.
"""
)

kmeans = KMeans(
    n_clusters=8,
    random_state=RANDOM_STATE,
    n_init=10
)

kmeans.fit(sales_df)

labels = kmeans.labels_

print("First 20 cluster labels:", labels[:20].tolist())
print("Unique cluster labels:", sorted(np.unique(labels).tolist()))
print("Inertia:", round(kmeans.inertia_, 2))


# ============================================================
# 5. ATTACH LABELS AND AGGREGATE SALES
# ============================================================

section(
    "5. Attach labels and aggregate sales",
    """
We create a DataFrame containing the original 52 weekly sales columns
and the cluster label generated by K-Means.

For each cluster, we calculate:
- count_product: number of products in the cluster
- total_sales: total units sold across all 52 weeks
- yearly_average_sales: total sales divided by the number of products

The results are sorted by yearly_average_sales in descending order,
as shown in the textbook example.
"""
)

clustered = df[weekly_cols].copy()

clustered["label"] = labels

clustered["annual_sales"] = clustered[weekly_cols].sum(axis=1)

summary = (
    clustered.groupby("label")
    .agg(
        count_product=("annual_sales", "size"),
        total_sales=("annual_sales", "sum")
    )
)

summary["yearly_average_sales"] = (
    summary["total_sales"] / summary["count_product"]
)

summary = summary.sort_values(
    "yearly_average_sales",
    ascending=False
)

display(summary)

print("Products accounted for:", summary["count_product"].sum())
print("Dataset annual sales total:", summary["total_sales"].sum())

assert summary["count_product"].sum() == len(df)

assert summary["total_sales"].sum() == (
    df[weekly_cols].to_numpy().sum()
)

display(Markdown("""
**Interpreting the summary:** A higher yearly average indicates that
products in that cluster sold more units across the 52-week period
on average. Cluster IDs are arbitrary and may differ between software
versions or initialization choices. The yearly average is the total
annual sales per product, not the average for one week.
"""))


# ============================================================
# 6. VISUALIZE THE EIGHT CLUSTERS
# ============================================================

section(
    "6. Visualize the eight clusters",
    """
The first bar chart compares average annual sales per product across
clusters. The second chart plots each cluster's mean weekly sales,
making differences in sales levels and weekly patterns easier to see.
"""
)

# Chart 1: Average annual sales per product
fig, ax = plt.subplots(figsize=(9, 4.5))

summary["yearly_average_sales"].plot.bar(ax=ax)

ax.set(
    title="Average Annual Sales per Product by Cluster",
    xlabel="Cluster label",
    ylabel="Units sold per product (52 weeks)"
)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# Chart 2: Average weekly sales profiles
weekly_profiles = clustered.groupby("label")[weekly_cols].mean()

fig, ax = plt.subplots(figsize=(10, 5))

for label, row in weekly_profiles.iterrows():
    ax.plot(
        range(52),
        row.values,
        label=f"Cluster {label}",
        linewidth=1.5
    )

ax.set(
    title="Average Weekly Sales Profiles for Eight Clusters",
    xlabel="Week (W0–W51)",
    ylabel="Average units sold"
)

ax.legend(ncol=2, fontsize=8)
plt.tight_layout()
plt.show()


# ============================================================
# 7. EXPLORE ALTERNATIVE CLUSTER COUNTS
# ============================================================

section(
    "7. Explore alternative cluster counts",
    """
We experiment with k values from 2 through 12, inclusive, using
inertia and the Silhouette Score.

Inertia measures within-cluster squared distances and generally
decreases as the number of clusters increases. A higher silhouette
score generally indicates more compact, better-separated groups.
Neither metric alone guarantees the most useful business grouping.
"""
)

candidate_k = list(range(2, 13))

inertias = []
silhouettes = []

for k in candidate_k:
    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=10
    )

    candidate_labels = model.fit_predict(sales_df)

    inertias.append(model.inertia_)

    silhouettes.append(
        silhouette_score(sales_df, candidate_labels)
    )

metrics = pd.DataFrame({
    "k": candidate_k,
    "inertia": inertias,
    "silhouette_score": silhouettes
})

display(metrics.round(4))

best_row = metrics.loc[metrics["silhouette_score"].idxmax()]

print(
    f"Best silhouette among tested values: "
    f"k={int(best_row['k'])}, "
    f"score={best_row['silhouette_score']:.4f}"
)

k8_score = metrics.loc[
    metrics["k"] == 8,
    "silhouette_score"
].iloc[0]

print(f"Silhouette score for required k=8: {k8_score:.4f}")


# Chart 3: Elbow method
fig, ax = plt.subplots(figsize=(8, 4.5))

ax.plot(
    metrics["k"],
    metrics["inertia"],
    marker="o"
)

ax.set(
    title="Elbow Method: Inertia vs. Number of Clusters",
    xlabel="Number of clusters (k)",
    ylabel="Inertia"
)

ax.set_xticks(candidate_k)
plt.tight_layout()
plt.show()


# Chart 4: Silhouette scores
fig, ax = plt.subplots(figsize=(8, 4.5))

ax.plot(
    metrics["k"],
    metrics["silhouette_score"],
    marker="o"
)

ax.set(
    title="Silhouette Score vs. Number of Clusters",
    xlabel="Number of clusters (k)",
    ylabel="Silhouette Score"
)

ax.set_xticks(candidate_k)
plt.tight_layout()
plt.show()


# ============================================================
# 8. CONCLUSION
# ============================================================

section(
    "8. Conclusion",
    """
The required K-Means model divides the products into eight groups
based on their 52-week normalized sales vectors. The group-level
table quantifies the number of products and their annual sales in
each cluster. The weekly profiles reveal whether groups differ
mainly in sales magnitude or also in their weekly patterns.

The elbow and silhouette analyses provide evidence for evaluating
other cluster counts, but the eight-cluster model remains the
assignment's required baseline.

**Limitation:** Normalized weekly values emphasize sales patterns
differently from raw sales. Cluster assignments and numeric labels
can differ from the textbook's historical software version, even
when the same workflow and random seed are used. This analysis uses
normalized values for clustering and original sales counts for
reported totals.
"""
)

print("Final dataset rows:", len(df))
print("Number of normalized clustering features:", sales_df.shape[1])
print("Required number of clusters:", 8)
print("Total products assigned to clusters:", int(summary["count_product"].sum()))
print("Total units sold over all 52 weeks:", int(summary["total_sales"].sum()))


# ============================================================
# REFERENCES
# ============================================================

display(Markdown("""
## References

So, A., So, W., Nagy, Z., Bagve, S., Doyle, J. W., Patil, A. P.,
Pathak, S. S., & Roy, S. (2020). *The Applied Artificial Intelligence
Workshop*. Packt Publishing.

Tan, S., & San Lau, J. (2014). Time series clustering: A superior
alternative for market basket analysis. In *Proceedings of the First
International Conference on Advanced Data and Information Engineering
(DaEng-2013)* (pp. 241–248).
"""))

print("Assignment 7.2 completed successfully.")
```
